```python

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from sklearn.datasets import load_iris
from sklearn.model_selection import KFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

HO_TEN = "TRẦN THỊ MỸ DUNG"
MSSV = "6451071008"

np.random.seed(42)
tf.random.set_seed(42)

def thong_tin(cau):
    print(f"Họ tên: {HO_TEN}")
    print(f"MSSV: {MSSV}")
    print(f"BÀI 4 - CÂU {cau}")
    print("-" * 50)

iris = load_iris()
X = iris.data.copy()
y = iris.target.copy()

max_values = np.max(X, axis=0)
X = X / max_values

print("Đọc và chuẩn hóa dữ liệu thành công!")
print("Kích thước dữ liệu:", X.shape)
print("Kích thước nhãn:", y.shape)
```

```python

  # Câu c: Xây dựng và biên dịch mô hình Neural Network
thong_tin("c")

def create_model():
      model = Sequential([
          Input(shape=(4,)),
          Dense(10, activation="relu"),
          Dense(20, activation="relu"),
          Dense(3, activation="softmax")
      ])

      model.compile(
          optimizer="adam",
          loss="sparse_categorical_crossentropy",
          metrics=["accuracy"]
      )
      return model

  model_demo = create_model()
  model_demo.summary()

  print("Đã tạo mô hình thành công!")
```

```python

# Câu d: Huấn luyện K-Fold và đánh giá từng fold
thong_tin("d")

kfold = KFold(n_splits=10, shuffle=True, random_state=42)

models = []
accuracy_per_fold = []
precision_per_fold = []
recall_per_fold = []
f1_per_fold = []

class_precision = []
class_recall = []
class_f1 = []

for fold_no, (train_idx, val_idx) in enumerate(kfold.split(X), start=1):
    print(f"\n========== FOLD {fold_no}/10 ==========")

    tf.keras.backend.clear_session()
    model = create_model()

    X_train_fold = X[train_idx]
    X_val_fold = X[val_idx]
    y_train_fold = y[train_idx]
    y_val_fold = y[val_idx]

    model.fit(
        X_train_fold,
        y_train_fold,
        epochs=100,
        batch_size=1,
        verbose=0
    )

    y_prob = model.predict(X_val_fold, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)

    acc = accuracy_score(y_val_fold, y_pred)
    pre = precision_score(
        y_val_fold, y_pred,
        average="macro", zero_division=0
    )
    rec = recall_score(
        y_val_fold, y_pred,
        average="macro", zero_division=0
    )
    f1 = f1_score(
        y_val_fold, y_pred,
        average="macro", zero_division=0
    )

    accuracy_per_fold.append(acc)
    precision_per_fold.append(pre)
    recall_per_fold.append(rec)
    f1_per_fold.append(f1)

    class_precision.append(
        precision_score(y_val_fold, y_pred, labels=[0,1,2],
                        average=None, zero_division=0)
    )
    class_recall.append(
        recall_score(y_val_fold, y_pred, labels=[0,1,2],
                     average=None, zero_division=0)
    )
    class_f1.append(
        f1_score(y_val_fold, y_pred, labels=[0,1,2],
                 average=None, zero_division=0)
    )

    models.append(model)

    print(f"Accuracy: {acc:.4f}")
    print(f"Precision: {pre:.4f}")
    print(f"Recall: {rec:.4f}")
    print(f"F1-score: {f1:.4f}")

print("\nHoàn thành huấn luyện 10 fold!")
```

```python

# Câu e: Tính trung bình Accuracy, Precision, Recall và F1-score
thong_tin("e")

print(f"Average Accuracy: {np.mean(accuracy_per_fold):.4f}")
print(f"Average Recall: {np.mean(recall_per_fold):.4f}")
print(f"Average Precision: {np.mean(precision_per_fold):.4f}")
print(f"Average F1-score: {np.mean(f1_per_fold):.4f}")
```

```python

# Câu f: Tính trung bình Precision, Recall và F1-score từng lớp
thong_tin("f")

avg_precision = np.mean(class_precision, axis=0)
avg_recall = np.mean(class_recall, axis=0)
avg_f1 = np.mean(class_f1, axis=0)

for i, class_name in enumerate(iris.target_names):
    print(f"\nLớp: {class_name}")
    print(f"Average Precision: {avg_precision[i]:.4f}")
    print(f"Average Recall: {avg_recall[i]:.4f}")
    print(f"Average F1-score: {avg_f1[i]:.4f}")
```

```python

# Câu g: Dùng mô hình có F1-score cao nhất dự đoán 3 mẫu Iris
thong_tin("g")

best_index = np.argmax(f1_per_fold)
best_model = models[best_index]

print("Fold có F1-score cao nhất:", best_index + 1)
print(f"F1-score cao nhất: {f1_per_fold[best_index]:.4f}")

new_samples = np.array([
    [6.2, 2.9, 4.3, 1.3],
    [5.1, 3.5, 1.4, 0.2],
    [7.3, 2.8, 6.4, 2.1]
])

new_samples_scaled = new_samples / max_values

predictions = best_model.predict(new_samples_scaled, verbose=0)
predicted_labels = np.argmax(predictions, axis=1)

for i, label in enumerate(predicted_labels):
    print(f"\nMẫu {i+1}: {new_samples[i]}")
    print("Nhãn dự đoán:", iris.target_names[label])
```
