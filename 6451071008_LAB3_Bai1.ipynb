```python
#Import thư viện
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense

np.random.seed(42)
tf.random.set_seed(42)

HO_TEN = "TRẦN THỊ MỸ DUNG"
MSSV = "6451071008"

def thong_tin(cau):
    print(f"Họ tên: {HO_TEN}")
    print(f"MSSV: {MSSV}")
    print(f"BÀI 1 - CÂU {cau}")
    print("-" * 50)

print("Đã import thư viện thành công!")
```

```python
# Câu a
thong_tin("a")

iris = load_iris()

X = iris.data
y = iris.target

print("Đọc dữ liệu Iris thành công!")
print("Kích thước dữ liệu:", X.shape)
print("Kích thước nhãn:", y.shape)
```

```python
# Câu b
thong_tin("b")

print("Tên các thuộc tính:")
print(iris.feature_names)

print("\nTên các lớp phân loại:")
print(iris.target_names)

print("\nSố lượng mẫu:", X.shape[0])
print("Số lượng thuộc tính:", X.shape[1])

print("\n5 mẫu dữ liệu đầu tiên:")
print(X[:5])

print("\nNhãn của 5 mẫu đầu tiên:")
print(y[:5])

df = pd.DataFrame(X, columns=iris.feature_names)

print("\nThống kê dữ liệu Iris:")
print(df.describe().to_string())
```

```python
# Câu c
thong_tin("c")

X = iris.data.copy()
Y = iris.target.copy()

max_values = np.max(X, axis=0)
X = X / max_values

print("Giá trị lớn nhất từng thuộc tính:")
print(max_values)

print("\n5 mẫu sau chuẩn hóa:")
print(X[:5])

print("\nGiá trị nhỏ nhất:", X.min())
print("Giá trị lớn nhất:", X.max())
```

```python
# Câu d
thong_tin("d")

X_train, X_test, y_train, y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42
)

print("Kích thước X_train:", X_train.shape)
print("Kích thước X_test:", X_test.shape)

print("Kích thước y_train:", y_train.shape)
print("Kích thước y_test:", y_test.shape)
```

```python
# Câu e
thong_tin("e - Decision Tree")

dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)

acc_dt = accuracy_score(y_test, y_pred_dt)

print("Accuracy:", acc_dt)

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_dt,
    target_names=iris.target_names,
    zero_division=0
))

cm_dt = confusion_matrix(y_test, y_pred_dt)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title("Confusion Matrix - Decision Tree")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```

```python
# Câu f
thong_tin("f - Naive Bayes")

nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

y_pred_nb = nb_model.predict(X_test)

acc_nb = accuracy_score(y_test, y_pred_nb)

print("Accuracy:", acc_nb)

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_nb,
    target_names=iris.target_names,
    zero_division=0
))

cm_nb = confusion_matrix(y_test, y_pred_nb)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_nb,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title("Confusion Matrix - Naive Bayes")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```

```python
# Câu g
thong_tin("g - K Nearest Neighbors")

knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

y_pred_knn = knn_model.predict(X_test)

acc_knn = accuracy_score(y_test, y_pred_knn)

print("Accuracy:", acc_knn)

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_knn,
    target_names=iris.target_names,
    zero_division=0
))

cm_knn = confusion_matrix(y_test, y_pred_knn)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_knn,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title("Confusion Matrix - KNN")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```

```python
# Câu h
thong_tin("h - Neural Network")

nn_model = Sequential([
    Input(shape=(4,)),
    Dense(10, activation="relu"),
    Dense(20, activation="relu"),
    Dense(3, activation="softmax")
])

nn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

nn_model.summary()

history = nn_model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=1,
    verbose=1
)

y_prob_nn = nn_model.predict(X_test)
y_pred_nn = np.argmax(y_prob_nn, axis=1)

acc_nn = accuracy_score(y_test, y_pred_nn)

print("\nAccuracy Neural Network:", acc_nn)

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_nn,
    target_names=iris.target_names,
    zero_division=0
))

cm_nn = confusion_matrix(y_test, y_pred_nn)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm_nn,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)

plt.title("Confusion Matrix - Neural Network")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```

```python
# Câu i
thong_tin("i - Du doan 3 mau moi")

new_samples = np.array([
    [6.2, 2.9, 4.3, 1.3],
    [5.1, 3.5, 1.4, 0.2],
    [7.3, 2.8, 6.4, 2.1]
])

# Chuẩn hóa theo giá trị max của dữ liệu gốc
new_samples_scaled = new_samples / max_values

pred_dt = dt_model.predict(new_samples_scaled)
pred_nb = nb_model.predict(new_samples_scaled)
pred_knn = knn_model.predict(new_samples_scaled)

pred_nn = np.argmax(
    nn_model.predict(new_samples_scaled),
    axis=1
)

for i in range(len(new_samples)):
    print(f"\nMẫu {i + 1}: {new_samples[i]}")

    print("Decision Tree:",
          iris.target_names[pred_dt[i]])

    print("Naive Bayes:",
          iris.target_names[pred_nb[i]])

    print("KNN:",
          iris.target_names[pred_knn[i]])

    print("Neural Network:",
          iris.target_names[pred_nn[i]])
```

```python
# Câu j
thong_tin("j - Danh gia cac mo hinh")

results = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Naive Bayes",
        "KNN",
        "Neural Network"
    ],
    "Accuracy": [
        acc_dt,
        acc_nb,
        acc_knn,
        acc_nn
    ]
})

results = results.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print("Bảng so sánh Accuracy:")
print(results.to_string(index=False))

plt.figure(figsize=(9, 5))
sns.barplot(
    data=results,
    x="Model",
    y="Accuracy",
    color="steelblue"
)

plt.title("So sanh Accuracy cua 4 mo hinh")
plt.ylim(0, 1.1)
plt.xticks(rotation=15)

for i, acc in enumerate(results["Accuracy"]):
    plt.text(i, acc + 0.02,
             f"{acc:.4f}",
             ha="center")

plt.tight_layout()
plt.show()

print("\nMô hình có Accuracy cao nhất:")
print(results.iloc[0]["Model"])

print("Accuracy cao nhất:",
      results.iloc[0]["Accuracy"])
```
