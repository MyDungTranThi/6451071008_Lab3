```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense, Dropout
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

HO_TEN = "TRẦN THỊ MỸ DUNG"
MSSV = "6451071008"

np.random.seed(42)
tf.random.set_seed(42)

def thong_tin(cau):
    print(f"Họ tên: {HO_TEN}")
    print(f"MSSV: {MSSV}")
    print(f"BÀI 2 - CÂU {cau}")
    print("-" * 50)

print("Import thư viện thành công!")
```

```python

# Câu a: Đọc bộ dữ liệu MNIST từ Keras
thong_tin("a")

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

print("Đọc dữ liệu MNIST thành công!")
print("Số ảnh train:", len(train_images))
print("Số ảnh test:", len(test_images))
```

```python

# Câu b: In kích thước tập train và test
thong_tin("b")

print("Train images:", train_images.shape)
print("Train labels:", train_labels.shape)
print("Test images:", test_images.shape)
print("Test labels:", test_labels.shape)
```

```python

# Câu c: Chuẩn hóa giá trị pixel về đoạn [0,1]
thong_tin("c")

train_images = train_images.astype("float32") / 255.0
test_images = test_images.astype("float32") / 255.0

print("Min train:", train_images.min())
print("Max train:", train_images.max())
print("Min test:", test_images.min())
print("Max test:", test_images.max())
```

```python

# Câu d: Chia train và validation theo tỷ lệ 8:2
thong_tin("d")

train_images, validation_images, train_labels, validation_labels = train_test_split(
    train_images,
    train_labels,
    test_size=0.2,
    random_state=42
)

print("Train images:", train_images.shape)
print("Train labels:", train_labels.shape)
print("Validation images:", validation_images.shape)
print("Validation labels:", validation_labels.shape)
print("Test images:", test_images.shape)
```

```python

# Câu e: Xây dựng Neural Network với Flatten, Dense và Dropout
thong_tin("e")

model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.2),
    Dense(10, activation="softmax")
])

model.summary()
```

```python

# Câu f: Biên dịch mô hình với Adam và Accuracy
thong_tin("f")

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Biên dịch mô hình thành công!")
print("Optimizer: Adam")
print("Loss: sparse_categorical_crossentropy")
print("Metric: accuracy")
```

```python

# Câu g: Huấn luyện Neural Network với 5 epochs
thong_tin("g")

history = model.fit(
    train_images,
    train_labels,
    epochs=5,
    batch_size=32,
    validation_data=(validation_images, validation_labels),
    verbose=1
)

print("\nHoàn thành huấn luyện!")
```

```python

# Câu h: Vẽ biểu đồ Accuracy và Loss sau mỗi epoch
thong_tin("h")

epochs = range(1, len(history.history["accuracy"]) + 1)

plt.figure(figsize=(8, 5))
plt.plot(epochs, history.history["accuracy"], marker="o", label="Train Accuracy")
plt.plot(epochs, history.history["val_accuracy"], marker="o", label="Validation Accuracy")
plt.title("Accuracy sau moi epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(epochs, history.history["loss"], marker="o", label="Train Loss")
plt.plot(epochs, history.history["val_loss"], marker="o", label="Validation Loss")
plt.title("Loss sau moi epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Câu i: Dự đoán nhãn chữ số trên tập test
thong_tin("i")

y_prob = model.predict(test_images)
y_pred = np.argmax(y_prob, axis=1)

print("Kích thước xác suất dự đoán:", y_prob.shape)
print("Kích thước nhãn dự đoán:", y_pred.shape)

print("\n10 nhãn dự đoán đầu tiên:")
print(y_pred[:10])

print("\n10 nhãn thực tế đầu tiên:")
print(test_labels[:10])
```

```python

# Câu j: Tính Accuracy của mô hình trên tập test
thong_tin("j")

accuracy = accuracy_score(test_labels, y_pred)

print("Accuracy trên tập test:", accuracy)
print(f"Accuracy (%): {accuracy * 100:.2f}%")
```

```python

# Câu k: Đánh giá Precision, Recall và F1-score
thong_tin("k")

report = classification_report(
    test_labels,
    y_pred,
    labels=np.arange(10),
    target_names=[str(i) for i in range(10)],
    zero_division=0
)

print("Classification Report:")
print(report)
```

```python

# Câu l: Hiển thị Confusion Matrix bằng Heatmap
thong_tin("l")

cm = confusion_matrix(test_labels, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=range(10),
    yticklabels=range(10)
)

plt.title("Confusion Matrix - MNIST")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```

```python

# Câu m: Hiển thị kết quả phân lớp 5 ảnh đầu tiên
thong_tin("m")

plt.figure(figsize=(15, 4))

for i in range(5):
    image = np.expand_dims(test_images[i], axis=0)
    prediction = model.predict(image, verbose=0)
    predicted_label = np.argmax(prediction, axis=1)[0]

    print(f"Ảnh {i + 1}: Thực tế = {test_labels[i]}, Dự đoán = {predicted_label}")

    plt.subplot(1, 5, i + 1)
    plt.imshow(test_images[i], cmap="gray")
    plt.title(f"True: {test_labels[i]}\nPred: {predicted_label}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

# Câu n: Hiển thị 5 ảnh đầu tiên mô hình phân lớp sai
thong_tin("n")

wrong_indices = np.where(y_pred != test_labels)[0]
first_5_wrong = wrong_indices[:5]

print("Tổng số ảnh dự đoán sai:", len(wrong_indices))

plt.figure(figsize=(15, 4))

for i, idx in enumerate(first_5_wrong):
    print(f"Ảnh {idx}: Thực tế = {test_labels[idx]}, Dự đoán = {y_pred[idx]}")

    plt.subplot(1, 5, i + 1)
    plt.imshow(test_images[idx], cmap="gray")
    plt.title(f"True: {test_labels[idx]}\nPred: {y_pred[idx]}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
